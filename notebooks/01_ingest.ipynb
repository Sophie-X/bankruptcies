## Store the Meta Data / Lookups: SCD Type 1

The lookups (`bronze.dim_*`) are reloaded in full on every run, with no history (SCD Type 1).

- **Why not Type 2:** SIC 2008 is a fixed classification. A code always means the same sector; CBS only rewords titles. No analysis needs the old wording.
- **Why `TRUNCATE` + append, not `if_exists='replace'`:** `replace` drops the table, which fails once a silver view depends on it.
- **Join on codes, never on titles:** final periods are never reloaded, so they keep the title CBS used at the time. If CBS renames a sector, a title join silently drops those rows. That is why the ingest maps `period_key` (and `sector_key`) from the lookups.

## Load & Store The Data

On every run, ask CBS for the status of each period (metadata). Compare it with what was loaded last time, and ONLY (re)load a period if:
- New: the period isn’t in your database yet
- Provisional at CBS now
- Status changed since last load (for figures that move from provisional -> definitief)

To track changes, we utilise a control table. Below we connect to the local Postgres database: bankruptcies -> load_periods table in the CONTROL schema (schemas: CONTROL, BRONZE, SILVER, GOLD). The control table tracks our own loads and stores status per period per load_period. 

### Connect to PostgreSQL

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine, URL

load_dotenv()

url = URL.create(
    drivername="postgresql+psycopg",
    username=os.getenv("PGUSER"),
    password=os.getenv("PGPASSWORD"),
    host=os.getenv("PGHOST"),
    port=int(os.getenv("PGPORT")),
    database=os.getenv("PGDATABASE"),
)

engine = create_engine(url)
```

```python
from sqlalchemy import text

TABLE_ID = '82244ENG'

source = pd.DataFrame(cbsodata.get_meta(TABLE_ID, 'Periods'))[['Key', 'Status']]
source.columns = ['period_key', 'status']
source['status'] = source['status'].fillna('').str.strip()

loaded = pd.read_sql('SELECT period_key, status FROM control.load_periods', engine)

compare = source.merge(loaded, on='period_key', how='left',
                       suffixes=('', '_loaded'), indicator=True)

is_new         = compare['_merge'] == 'left_only'
is_provisional = compare['status'] != 'Definitief'
status_changed = ~is_new & (compare['status'] != compare['status_loaded'])

to_load = compare.loc[is_new | is_provisional | status_changed, 'period_key'].tolist()

print(f'New: {is_new.sum()} | Provisional: {is_provisional.sum()} | '
      f'Status changed: {status_changed.sum()} | To load: {len(to_load)} of {len(source)}')
```

```python
def fetch_periods(keys, batch_size=12):
    if loaded.empty:                       # first run: one full download is faster
        return pd.DataFrame(cbsodata.get_data(TABLE_ID))
    parts = []
    for i in range(0, len(keys), batch_size):
        batch = keys[i:i + batch_size]
        flt = ' or '.join(f"Periods eq '{k}'" for k in batch)
        parts.append(pd.DataFrame(cbsodata.get_data(TABLE_ID, filters=flt)))
    return pd.concat(parts, ignore_index=True)

if to_load:
    new_rows = fetch_periods(to_load)
    new_rows.columns = new_rows.columns.str.lower()
    new_rows['period_key'] = new_rows['periods'].map(dict(zip(periods_df['Title'], periods_df['Key'])))
    print(f'Downloaded {len(new_rows)} rows')
else:
    print('Nothing to load')
```

```python
from sqlalchemy import inspect

with engine.begin() as conn:
    if inspect(conn).has_table('raw_bankruptcies', schema='bronze'):
        conn.execute(text('DELETE FROM bronze.raw_bankruptcies WHERE period_key = ANY(:keys)'),
                     {'keys': to_load})
    new_rows.to_sql('raw_bankruptcies', conn, schema='bronze', if_exists='append', index=False)

    conn.execute(text('DELETE FROM control.load_periods WHERE period_key = ANY(:keys)'),
                 {'keys': to_load})
    log = source[source['period_key'].isin(to_load)].assign(loaded_at=pd.Timestamp.now(tz='UTC'))
    log.to_sql('load_periods', conn, schema='control', if_exists='append', index=False)
```
